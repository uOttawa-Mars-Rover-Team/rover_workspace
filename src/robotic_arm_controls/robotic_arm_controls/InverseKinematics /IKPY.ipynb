```python
import ikpy.chain
import ikpy.utils.plot as plot_utils

import numpy as np
import time
import math

import ipywidgets as widgets
import serial
import matplotlib.pyplot as plt
```

```python
my_chain = ikpy.chain.Chain.from_urdf_file("ikpytest.urdf",
                                            base_elements=["base_footprint"],
                                            active_links_mask=[False, False, True, True, True]
                                       )

# my_chain = ikpy.chain.Chain.from_urdf_file("rotechnicExURDF.urdf",active_links_mask=[False, True, True, True, True, True, True])
```

Set target position/orientation and compute an IK solution

```python
target_position= [0,0.2,0.1]
target_orientation=[-1,0,0]

ik = my_chain.inverse_kinematics(target_position, target_orientation)
```

Visualize the URDF's current position

```python
fig, ax = plot_utils.init_3d_figure()
fig.set_figheight(9)
fig.set_figwidth(13)

n = len(my_chain.links)
zeros = [0] * n
my_chain.plot(zeros, ax)

my_chain.plot(ik, ax, target=target_position)

plt.xlim(-0.5, 0.5)
plt.ylim(-0.5, 0.5)
ax.set_zlim(0, 0.6)
plt.ion()
```
